# **Part 1. Descriptive Analysis on IBM HR Data**

## **1. Importing Libraries and Data files**

```python
import pandas as pd

#Getting the source data and display it
url = "https://raw.githubusercontent.com/GJebres/LHUBCapstoneProject/main/SourceData/"

#Reading the files
compensation = pd.read_csv(url + "Compensation.csv")
employees = pd.read_csv(url + "Employees.csv")
employment_status = pd.read_csv(url + "EmploymentStatus.csv")
experience = pd.read_csv(url + "Experience.csv")
performance = pd.read_csv(url + "Performance.csv")

print("COMPENSATION")
print(compensation.shape)
print(compensation.columns.tolist())
display(compensation.head(3))

print("EMPLOYEES")
print(employees.shape)
print(employees.columns.tolist())
display(employees.head(3))

print("EMPLOYMENT STATUS")
print(employment_status.shape)
print(employment_status.columns.tolist())
display(employment_status.head(3))

print("EXPERIENCE")
print(experience.shape)
print(experience.columns.tolist())
display(experience.head(3))

print("PERFORMANCE")
print(performance.shape)
print(performance.columns.tolist())
display(performance.head(3))
```

## **3. Employee Attrition Analysis (RO1)**

```python
# Count employees who left and stayed
print(employment_status["Attrition"].value_counts())
```

```python
# Calculate attrition percentages
print(
    employment_status["Attrition"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

## **4. Department and Job Role Attrition Analysis (RO2)**

### **4.1. Analysis by Department**

```python
# Combine employee data with employment status
hr_data = pd.merge(
    employees,
    employment_status,
    on="EmployeeNumber",
    how="left"
)

# Check the combined data
print("Rows and columns:", hr_data.shape)
display(hr_data.head())
```

```python
# Count employees who left and stayed by department
department_attrition = pd.crosstab(
    hr_data["Department"],
    hr_data["Attrition"]
)

# Calculate attrition rate
department_attrition["Attrition Rate (%)"] = (
    department_attrition["Yes"] /
    (department_attrition["Yes"] + department_attrition["No"])
) * 100

print(department_attrition.round(2))
```

```python
import matplotlib.pyplot as plt

department_attrition["Attrition Rate (%)"].plot(
    kind="bar",
    title="Attrition Rate by Department",
    ylabel="Attrition Rate (%)",
    xlabel="Department",
    rot=0
)

plt.tight_layout()
plt.show()
```

### **4.2. Analysis by Job Role**

```python
# Count employees who left and stayed by job role
job_attrition = pd.crosstab(
    hr_data["JobRole"],
    hr_data["Attrition"]
)

# Calculate attrition rate
job_attrition["Attrition Rate (%)"] = (
    job_attrition["Yes"] /
    (job_attrition["Yes"] + job_attrition["No"])
) * 100

# Sort from highest to lowest
job_attrition = job_attrition.sort_values(
    "Attrition Rate (%)",
    ascending=False
)

print(job_attrition.round(2))
```

```python
import matplotlib.pyplot as plt

job_attrition["Attrition Rate (%)"].plot(
    kind="bar",
    title="Attrition Rate by Job Role",
    ylabel="Attrition Rate (%)",
    xlabel="Job Role"
)

plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

### **4.3. Analysis by Age group**

```python
# Create age groups
hr_data["Age Group"] = pd.cut(
    hr_data["Age"],
    bins=[0, 25, 35, 45, 55, 100],
    labels=["Under 26", "26-35", "36-45", "46-55", "56 and above"]
)

# Count employees who left and stayed
age_attrition = pd.crosstab(
    hr_data["Age Group"],
    hr_data["Attrition"]
)

# Calculate attrition rate
age_attrition["Attrition Rate (%)"] = (
    age_attrition["Yes"] /
    (age_attrition["Yes"] + age_attrition["No"])
) * 100

print(age_attrition.round(2))
```

```python
import matplotlib.pyplot as plt

age_attrition["Attrition Rate (%)"].plot(
    kind="bar",
    title="Attrition Rate by Age Group",
    ylabel="Attrition Rate (%)",
    xlabel="Age Group",
    rot=0
)

plt.tight_layout()
plt.show()
```

### **4.4. Analysis by Overtime status**

```python
# Count employees who left and stayed by overtime status
overtime_attrition = pd.crosstab(
    hr_data["OverTime"],
    hr_data["Attrition"]
)

# Calculate attrition rate
overtime_attrition["Attrition Rate (%)"] = (
    overtime_attrition["Yes"] /
    (overtime_attrition["Yes"] + overtime_attrition["No"])
) * 100

print(overtime_attrition.round(2))
```

```python
import matplotlib.pyplot as plt

overtime_attrition["Attrition Rate (%)"].plot(
    kind="bar",
    title="Attrition Rate by Overtime Status",
    ylabel="Attrition Rate (%)",
    xlabel="Works Overtime",
    rot=0
)

plt.tight_layout()
plt.show()
```

Summary of the findings based on the 4 analysis (Department, Job Role, Age Group and Overtime status)
