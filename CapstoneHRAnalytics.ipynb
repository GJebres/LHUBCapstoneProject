# **Part 2. Predictive analysis on IBM HR Data**

```python
print("Dataset size:", hr_data.shape)

print("\nColumn names:")
print(hr_data.columns.tolist())

print("\nMissing values:")
print(hr_data.isnull().sum().sum())

print("\nAttrition counts:")
print(hr_data["Attrition"].value_counts())

print("\nAttrition percentages:")
print(
    hr_data["Attrition"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python
# Separate input features and target variable

X = hr_data.drop(
    columns=["Attrition", "EmployeeNumber"],
    errors="ignore"
)

y = hr_data["Attrition"].astype(str).str.strip().str.title()

print("Input features:", X.shape)
print("\nTarget distribution:")
print(y.value_counts())

print("\nTarget values:")
print(y.unique())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training records:", X_train.shape[0])
print("Testing records:", X_test.shape[0])
print("Training features:", X_train.shape[1])
print("Testing features:", X_test.shape[1])
```

```python
import pandas as pd

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

# Identify numerical and categorical columns
numeric_features = X_train.select_dtypes(
    include=["number"]
).columns.tolist()

categorical_features = X_train.select_dtypes(
    exclude=["number"]
).columns.tolist()

print("Number of numerical features:", len(numeric_features))
print("Number of categorical features:", len(categorical_features))

# Prepare numerical data
numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Prepare categorical data
categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# Combine both preprocessing methods
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)

print("Data preprocessing pipeline created successfully.")
```

```python
# Model 1: Logistic Regression
logistic_model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", LogisticRegression(
        max_iter=2000,
        class_weight="balanced",
        random_state=42
    ))
])

# Model 2: Random Forest
random_forest_model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(
        n_estimators=300,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    ))
])

# Train both models using training data only
logistic_model.fit(X_train, y_train)
random_forest_model.fit(X_train, y_train)

print("Logistic Regression model trained successfully.")
print("Random Forest model trained successfully.")
```

```python
import pandas as pd

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

models = {
    "Logistic Regression": logistic_model,
    "Random Forest": random_forest_model
}

results = []

for name, model in models.items():
    predictions = model.predict(X_test)

    # Probability that an employee will leave
    yes_index = list(model.classes_).index("Yes")
    probabilities = model.predict_proba(X_test)[:, yes_index]

    # Calculate performance metrics
    results.append({
        "Model": name,
        "Accuracy (%)": round(
            accuracy_score(y_test, predictions) * 100, 2
        ),
        "Precision (%)": round(
            precision_score(
                y_test, predictions,
                pos_label="Yes", zero_division=0
            ) * 100, 2
        ),
        "Recall (%)": round(
            recall_score(
                y_test, predictions,
                pos_label="Yes", zero_division=0
            ) * 100, 2
        ),
        "F1-Score (%)": round(
            f1_score(
                y_test, predictions,
                pos_label="Yes", zero_division=0
            ) * 100, 2
        ),
        "ROC-AUC": round(
            roc_auc_score(
                (y_test == "Yes").astype(int),
                probabilities
            ), 3
        )
    })

    print("\n" + "=" * 45)
    print(name)
    print("=" * 45)

    print("\nClassification Report:")
    print(classification_report(
        y_test,
        predictions,
        labels=["No", "Yes"],
        zero_division=0
    ))

    print("\nConfusion Matrix (actual rows, predicted columns):")
    cm = confusion_matrix(
        y_test,
        predictions,
        labels=["No", "Yes"]
    )
    print(cm)

results_df = pd.DataFrame(results)

print("\nMODEL COMPARISON")
display(results_df)
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay

# Chart 1: Compare model performance
metrics = [
    "Accuracy (%)",
    "Precision (%)",
    "Recall (%)",
    "F1-Score (%)"
]

results_df.set_index("Model")[metrics].plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Logistic Regression vs Random Forest")
plt.ylabel("Score (%)")
plt.xlabel("Predictive Model")
plt.ylim(0, 100)
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# Chart 2: Confusion matrices
for name, model in models.items():
    ConfusionMatrixDisplay.from_estimator(
        model,
        X_test,
        y_test,
        labels=["No", "Yes"],
        display_labels=["Stayed", "Left"],
        cmap="Blues",
        values_format="d"
    )
    plt.title(f"Confusion Matrix - {name}")
    plt.tight_layout()
    plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.inspection import permutation_importance
from sklearn.metrics import f1_score, make_scorer

# Tell the scoring function that "Yes" is the positive class
attrition_f1 = make_scorer(
    f1_score,
    pos_label="Yes",
    zero_division=0
)

# Calculate permutation importance
importance = permutation_importance(
    random_forest_model,
    X_test,
    y_test,
    scoring=attrition_f1,
    n_repeats=10,
    random_state=42,
    n_jobs=-1
)

# Create the feature importance table
importance_df = pd.DataFrame({
    "Feature": X_test.columns,
    "Importance": importance.importances_mean,
    "Std_Deviation": importance.importances_std
}).sort_values(
    by="Importance",
    ascending=False
)

# Display the top 15 features
print("TOP 15 FEATURES FOR ATTRITION PREDICTION")
display(importance_df.head(15).round(4))

# Plot the top 10 features
top_features = importance_df.head(10).sort_values(
    by="Importance",
    ascending=True
)

plt.figure(figsize=(10, 6))
plt.barh(
    top_features["Feature"],
    top_features["Importance"]
)
plt.axvline(x=0, linewidth=1)
plt.title("Top 10 Features Influencing Random Forest Predictions")
plt.xlabel("Decrease in F1-score when feature is shuffled")
plt.ylabel("Employee Feature")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd

from sklearn.base import clone
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score
)

# Split the training data into training and validation sets
X_subtrain, X_val, y_subtrain, y_val = train_test_split(
    X_train,
    y_train,
    test_size=0.20,
    random_state=42,
    stratify=y_train
)

# Make separate copies of the existing model pipelines
lr_tuning = clone(logistic_model)
rf_tuning = clone(random_forest_model)

# Train on the smaller training set only
lr_tuning.fit(X_subtrain, y_subtrain)
rf_tuning.fit(X_subtrain, y_subtrain)

# Thresholds to compare
thresholds = [0.20, 0.30, 0.40, 0.50]

tuning_results = []

for name, model in {
    "Logistic Regression": lr_tuning,
    "Random Forest": rf_tuning
}.items():

    # Probability of an employee leaving
    yes_index = list(model.classes_).index("Yes")
    probabilities = model.predict_proba(X_val)[:, yes_index]

    for threshold in thresholds:
        # Apply the selected threshold
        predictions = [
            "Yes" if p >= threshold else "No"
            for p in probabilities
        ]

        tuning_results.append({
            "Model": name,
            "Threshold": threshold,
            "Precision (%)": round(
                precision_score(
                    y_val, predictions,
                    pos_label="Yes", zero_division=0
                ) * 100, 2
            ),
            "Recall (%)": round(
                recall_score(
                    y_val, predictions,
                    pos_label="Yes", zero_division=0
                ) * 100, 2
            ),
            "F1-Score (%)": round(
                f1_score(
                    y_val, predictions,
                    pos_label="Yes", zero_division=0
                ) * 100, 2
            )
        })

tuning_df = pd.DataFrame(tuning_results)

print("VALIDATION RESULTS BY MODEL AND THRESHOLD")
display(tuning_df)
```

```python
import matplotlib.pyplot as plt

for metric in ["Recall (%)", "Precision (%)", "F1-Score (%)"]:
    pivot = tuning_df.pivot(
        index="Threshold",
        columns="Model",
        values=metric
    )

    pivot.plot(
        marker="o",
        figsize=(8, 5)
    )

    plt.title(f"{metric} at Different Prediction Thresholds")
    plt.ylabel(metric)
    plt.xlabel("Prediction Threshold")
    plt.ylim(0, 100)
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()
```
