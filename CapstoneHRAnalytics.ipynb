Part 1. Descriptive Analysis on IBM HR Data

```python
import pandas as pd

#Getting the source data and display it
url = "https://raw.githubusercontent.com/GJebres/LHUBCapstoneProject/main/SourceData/"

#Reading the files
compensation = pd.read_csv(url + "Compensation.csv")
employees = pd.read_csv(url + "Employees.csv")
employment_status = pd.read_csv(url + "EmploymentStatus.csv")
experience = pd.read_csv(url + "Experience.csv")
performance = pd.read_csv(url + "Performance.csv")

print("COMPENSATION")
print(compensation.shape)
print(compensation.columns.tolist())
display(compensation.head(3))

print("EMPLOYEES")
print(employees.shape)
print(employees.columns.tolist())
display(employees.head(3))

print("EMPLOYMENT STATUS")
print(employment_status.shape)
print(employment_status.columns.tolist())
display(employment_status.head(3))

print("EXPERIENCE")
print(experience.shape)
print(experience.columns.tolist())
display(experience.head(3))

print("PERFORMANCE")
print(performance.shape)
print(performance.columns.tolist())
display(performance.head(3))
```

```python
#Check if there are missing values
print("Compensation")
print(compensation.isnull().sum())

print("Employees")
print(employees.isnull().sum())

print("Employment Status")
print(employment_status.isnull().sum())

print("Experience")
print(experience.isnull().sum())

print("Performance")
print(performance.isnull().sum())
```

```python
#Check if there are duplicate rows
print("Compensation duplicates:", compensation.duplicated().sum())
print("Employees duplicates:", employees.duplicated().sum())
print("Employment Status duplicates:", employment_status.duplicated().sum())
print("Experience duplicates:", experience.duplicated().sum())
print("Performance duplicates:", performance.duplicated().sum())
```

```python
#Check the data information
print("Compensation")
compensation.info()

print("Employees")
employees.info()

print("Employment Status")
employment_status.info()

print("Experience")
experience.info()

print("Performance")
performance.info()
```
