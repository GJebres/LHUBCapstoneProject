## **4. Department and Job Role Attrition Analysis (RO2)**

### **4.1. Analysis by Department**

```python
# Combine employee data with employment status
hr_data = pd.merge(
    employees,
    employment_status,
    on="EmployeeNumber",
    how="left"
)

# Check the combined data
print("Rows and columns:", hr_data.shape)
display(hr_data.head())
```

```python
# Count employees who left and stayed by department
department_attrition = pd.crosstab(
    hr_data["Department"],
    hr_data["Attrition"]
)

# Calculate attrition rate
department_attrition["Attrition Rate (%)"] = (
    department_attrition["Yes"] /
    (department_attrition["Yes"] + department_attrition["No"])
) * 100

print(department_attrition.round(2))
```

```python
import matplotlib.pyplot as plt

department_attrition["Attrition Rate (%)"].plot(
    kind="bar",
    title="Attrition Rate by Department",
    ylabel="Attrition Rate (%)",
    xlabel="Department",
    rot=0
)

plt.tight_layout()
plt.show()
```

### **4.2. Analysis by Job Role**

```python
# Count employees who left and stayed by job role
job_attrition = pd.crosstab(
    hr_data["JobRole"],
    hr_data["Attrition"]
)

# Calculate attrition rate
job_attrition["Attrition Rate (%)"] = (
    job_attrition["Yes"] /
    (job_attrition["Yes"] + job_attrition["No"])
) * 100

# Sort from highest to lowest
job_attrition = job_attrition.sort_values(
    "Attrition Rate (%)",
    ascending=False
)

print(job_attrition.round(2))
```

```python
import matplotlib.pyplot as plt

job_attrition["Attrition Rate (%)"].plot(
    kind="bar",
    title="Attrition Rate by Job Role",
    ylabel="Attrition Rate (%)",
    xlabel="Job Role"
)

plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

### **4.4. Analysis by Overtime status**

```python
# Count employees who left and stayed by overtime status
overtime_attrition = pd.crosstab(
    hr_data["OverTime"],
    hr_data["Attrition"]
)

# Calculate attrition rate
overtime_attrition["Attrition Rate (%)"] = (
    overtime_attrition["Yes"] /
    (overtime_attrition["Yes"] + overtime_attrition["No"])
) * 100

print(overtime_attrition.round(2))
```

```python
import pandas as pd

# 1. Job satisfaction
# 2. Total working years
# 3. Experience groups
# 4. Years at company

# Helper function to summarise attrition
def attrition_summary(df, group_col):

    result = df.groupby(
    group_col,
    dropna=False,
    observed=False
).agg(
    Total_Employees=("Attrition", "size"),
    Employees_Left=("Attrition", lambda x: (x == "Yes").sum())
)

    result["Attrition_Rate_%"] = (
        result["Employees_Left"] /
        result["Total_Employees"] * 100
    )

    return result.round(2)


# A. Job satisfaction
print("\nJOB SATISFACTION")
print(attrition_summary(hr_data, "JobSatisfaction"))


# B. Total working years: compare employees who left and stayed
print("\nAVERAGE TOTAL WORKING YEARS")
print(
    hr_data.groupby("Attrition")["TotalWorkingYears"]
    .mean()
    .round(2)
)


# C. Experience groups
# Adjust these groups if your project uses different ranges
hr_data["ExperienceGroup"] = pd.cut(
    hr_data["TotalWorkingYears"],
    bins=[-1, 5, 10, 20, float("inf")],
    labels=["0–5 years", "6–10 years", "11–20 years", "Over 20 years"]
)

print("\nEXPERIENCE GROUPS")
print(attrition_summary(hr_data, "ExperienceGroup"))


# D. Years at company: compare employees who left and stayed
print("\nAVERAGE YEARS AT COMPANY")
print(
    hr_data.groupby("Attrition")["YearsAtCompany"]
    .mean()
    .round(2)
)


# E. Service groups
hr_data["ServiceGroup"] = pd.cut(
    hr_data["YearsAtCompany"],
    bins=[-1, 2, 5, 10, float("inf")],
    labels=["0–2 years", "3–5 years", "6–10 years", "Over 10 years"]
)

print("\nSERVICE GROUPS")
print(attrition_summary(hr_data, "ServiceGroup"))
```
