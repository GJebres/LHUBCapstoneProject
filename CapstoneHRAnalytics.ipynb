## **3. Research Objectives**

## **Research Objectives** ##
1. **Employee Attrition Rate and Workforce Analysis:** <br/> To measure the overall employee attrition rate and compare employees who leave with those who remain in the organisation.

```python
# Combine the five datasets using EmployeeNumber

hr_data = employees.merge(
    compensation,
    on="EmployeeNumber",
    how="left"
)

hr_data = hr_data.merge(
    employment_status,
    on="EmployeeNumber",
    how="left"
)

hr_data = hr_data.merge(
    experience,
    on="EmployeeNumber",
    how="left"
)

hr_data = hr_data.merge(
    performance,
    on="EmployeeNumber",
    how="left"
)

# Check the combined dataset
print("Number of rows and columns:", hr_data.shape)
print("Column names:", hr_data.columns.tolist())
print("Unique employees:", hr_data["EmployeeNumber"].nunique())

display(hr_data.head())

hr_data.to_csv('hr_data.csv', index=False)

import pandas as pd
from google.colab import files

files.download('hr_data.csv')
```

## **3.2. Department and Job Role Attrition Analysis (RO2)**

## **3.2.1. Attrition Rate by Department** ##

```python
# Count employees who left and stayed by department
department_attrition = pd.crosstab(
    hr_data["Department"],
    hr_data["Attrition"]
)

# Calculate attrition rate
department_attrition["Attrition Rate (%)"] = (
    department_attrition["Yes"] /
    (department_attrition["Yes"] + department_attrition["No"])
) * 100

print(department_attrition.round(2))

import matplotlib.pyplot as plt

department_attrition["Attrition Rate (%)"].plot(
    kind="bar",
    title="Attrition Rate by Department",
    ylabel="Attrition Rate (%)",
    xlabel="Department",
    rot=0
)

plt.tight_layout()
plt.show()
```

## **Observation** ##
Key observations:
- Sales has the highest attrition rate at 20.63%, which means approximately 21 out of every 100 employees in this department left.
- Human Resources has the second-highest rate at 19.05%.
- Research & Development has the lowest rate at 13.84%, although it has the largest number of employees who left (133).



## **3.2.2. Attrition Rate by Job Role**

```python
# Count employees who left and stayed by job role
job_attrition = pd.crosstab(
    hr_data["JobRole"],
    hr_data["Attrition"]
)

# Calculate attrition rate
job_attrition["Attrition Rate (%)"] = (
    job_attrition["Yes"] /
    (job_attrition["Yes"] + job_attrition["No"])
) * 100

# Sort from highest to lowest
job_attrition = job_attrition.sort_values(
    "Attrition Rate (%)",
    ascending=False
)

print(job_attrition.round(2))

import matplotlib.pyplot as plt

job_attrition["Attrition Rate (%)"].plot(
    kind="bar",
    title="Attrition Rate by Job Role",
    ylabel="Attrition Rate (%)",
    xlabel="Job Role"
)

plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

## **Observation** ##

Key findings:
- Sales Representatives have the highest attrition rate at 39.76%. Nearly 4 in 10 employees in this role left.
- Laboratory Technicians have the second-highest rate at 23.94%, followed by Human Resources employees at 23.08%.
- Research Directors have the lowest attrition rate at 2.50%, followed by Managers at 4.90%.
- Laboratory Technicians have the highest number of employees who left (62), even though Sales Representatives have the highest attrition rate.



## **3.3. Compensation and Employee Attrition Analysis (RO3)** ##

```python
# RO3: Compare compensation by attrition status

compensation_summary = (
    hr_data.groupby("Attrition")[
        [
            "MonthlyIncome",
            "PercentSalaryHike",
            "StockOptionLevel"
        ]
    ]
    .mean()
    .round(2)
)

display(compensation_summary)

# Group employees into monthly income bands

hr_data["Income Band"] = pd.cut(
    hr_data["MonthlyIncome"],
    bins=[0, 3000, 5000, 8000, float("inf")],
    labels=["Below 3,000", "3,000–4,999",
            "5,000–7,999", "8,000 and above"]
)

income_attrition = pd.crosstab(
    hr_data["Income Band"],
    hr_data["Attrition"],
    normalize="index"
) * 100

income_attrition["Attrition Rate (%)"] = (
    income_attrition["Yes"].round(2)
)

display(income_attrition.round(2))

hr_data.to_csv("combined_hr_data.csv", index=False)

print("CSV file created successfully.")
print("Rows:", hr_data.shape[0])
print("Columns:", hr_data.shape[1])
```

```python
def attrition_summary(df, column):
    summary = (
        df.groupby(column, observed=False)["Attrition"]
        .value_counts()
        .unstack(fill_value=0)
    )

    summary["Total Employees"] = summary.sum(axis=1)

    summary["Employees Left"] = summary.get("Yes", 0)

    summary["Attrition Rate (%)"] = (
        summary["Employees Left"]
        / summary["Total Employees"]
        * 100
    ).round(2)

    return summary
```

```python
# A. Job satisfaction
print("\nJOB SATISFACTION")
print(attrition_summary(hr_data, "JobSatisfaction"))


# B. Environment satisfaction
print("\nENVIRONMENT SATISFACTION")
print(attrition_summary(hr_data, "EnvironmentSatisfaction"))


# C. Work-life balance
print("\nWORK-LIFE BALANCE")
print(attrition_summary(hr_data, "WorkLifeBalance"))


# D. Total working years: compare employees who left and stayed
print("\nAVERAGE TOTAL WORKING YEARS")
print(
    hr_data.groupby("Attrition")["TotalWorkingYears"]
    .mean()
    .round(2)
)


# E. Experience groups
hr_data["ExperienceGroup"] = pd.cut(
    hr_data["TotalWorkingYears"],
    bins=[-1, 5, 10, 20, float("inf")],
    labels=["0–5 years", "6–10 years", "11–20 years", "Over 20 years"]
)

print("\nEXPERIENCE GROUPS")
print(attrition_summary(hr_data, "ExperienceGroup"))


# F. Years at company: compare employees who left and stayed
print("\nAVERAGE YEARS AT COMPANY")
print(
    hr_data.groupby("Attrition")["YearsAtCompany"]
    .mean()
    .round(2)
)


# G. Service groups
hr_data["ServiceGroup"] = pd.cut(
    hr_data["YearsAtCompany"],
    bins=[-1, 2, 5, 10, float("inf")],
    labels=["0–2 years", "3–5 years", "6–10 years", "Over 10 years"]
)

print("\nSERVICE GROUPS")
print(attrition_summary(hr_data, "ServiceGroup"))
```

```python
# Separate input features and target variable

X = hr_data.drop(
    columns=["Attrition", "EmployeeNumber"],
    errors="ignore"
)

y = hr_data["Attrition"].astype(str).str.strip().str.title()

print("Input features:", X.shape)
print("\nTarget distribution:")
print(y.value_counts())

print("\nTarget values:")
print(y.unique())
```

```python
# Define the target variable
y = hr_data["Attrition"]

# Define the input features
X = hr_data.drop(
    columns=[
        "Attrition",
        "EmployeeNumber",
        "Income Band",
        "ExperienceGroup",
        "ServiceGroup"
    ]
)

# Check the input features and target
print("Input features:", X.shape)
print("\nTarget distribution:")
print(y.value_counts())

print("\nTarget values:")
print(y.unique())
```

```python
from sklearn.model_selection import train_test_split

# Split the dataset: 80% training, 20% testing
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)

print("\nTraining target distribution:")
print(y_train.value_counts())

print("\nTesting target distribution:")
print(y_test.value_counts())
```

```python
# Identify numerical and categorical features
numerical_features = X_train.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = X_train.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

print("Number of numerical features:", len(numerical_features))
print(numerical_features)

print("\nNumber of categorical features:", len(categorical_features))
print(categorical_features)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline

# Remove columns that provide no useful predictive information
columns_to_drop = [
    "EmployeeCount",
    "StandardHours",
    "Over18"
]

X_train_model = X_train.drop(columns=columns_to_drop, errors="ignore")
X_test_model = X_test.drop(columns=columns_to_drop, errors="ignore")

# Identify numerical and categorical columns again
numerical_features = X_train_model.select_dtypes(
    include=["number"]
).columns.tolist()

categorical_features = X_train_model.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

# Prepare numerical features
numerical_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Prepare categorical features
categorical_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# Combine both preprocessing pipelines
preprocessor = ColumnTransformer(transformers=[
    ("num", numerical_pipeline, numerical_features),
    ("cat", categorical_pipeline, categorical_features)
])

print("Numerical features:", len(numerical_features))
print("Categorical features:", len(categorical_features))
print("Training records:", X_train_model.shape[0])
print("Testing records:", X_test_model.shape[0])
```

```python
from sklearn.linear_model import LogisticRegression

# Create the Logistic Regression pipeline
logistic_model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", LogisticRegression(
        max_iter=2000,
        class_weight="balanced",
        random_state=42
    ))
])

# Train the model
logistic_model.fit(X_train_model, y_train)

# Predict attrition on the test set
y_pred_logistic = logistic_model.predict(X_test_model)

# Predict probability of attrition = Yes
y_prob_logistic = logistic_model.predict_proba(
    X_test_model
)[:, list(logistic_model.named_steps["classifier"].classes_).index("Yes")]

print("Logistic Regression model trained successfully.")
print("Number of test predictions:", len(y_pred_logistic))
```

```python
import pandas as pd

from sklearn.base import clone
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score
)

# Split the training data into training and validation sets
X_subtrain, X_val, y_subtrain, y_val = train_test_split(
    X_train,
    y_train,
    test_size=0.20,
    random_state=42,
    stratify=y_train
)

# Make separate copies of the existing model pipelines
lr_tuning = clone(logistic_model)
rf_tuning = clone(random_forest_model)

# Train on the smaller training set only
lr_tuning.fit(X_subtrain, y_subtrain)
rf_tuning.fit(X_subtrain, y_subtrain)

# Thresholds to compare
thresholds = [0.20, 0.30, 0.40, 0.50]

tuning_results = []

for name, model in {
    "Logistic Regression": lr_tuning,
    "Random Forest": rf_tuning
}.items():

    # Get probability of an employee leaving
    yes_index = list(model.classes_).index("Yes")
    probabilities = model.predict_proba(X_val)[:, yes_index]

    for threshold in thresholds:

        # Apply the classification threshold
        predictions = [
            "Yes" if p >= threshold else "No"
            for p in probabilities
        ]

        # Calculate evaluation metrics
        precision = precision_score(
            y_val, predictions,
            pos_label="Yes",
            zero_division=0
        )

        recall = recall_score(
            y_val, predictions,
            pos_label="Yes",
            zero_division=0
        )

        f1 = f1_score(
            y_val, predictions,
            pos_label="Yes",
            zero_division=0
        )

        # Store the results
        tuning_results.append({
            "Model": name,
            "Threshold": threshold,
            "Precision (%)": round(precision * 100, 2),
            "Recall (%)": round(recall * 100, 2),
            "F1-Score (%)": round(f1 * 100, 2),
            "Predicted Attrition Cases": sum(
                p == "Yes" for p in predictions
            )
        })

# Display results
tuning_df = pd.DataFrame(tuning_results)

print("VALIDATION RESULTS BY MODEL AND THRESHOLD")
display(tuning_df)
```
